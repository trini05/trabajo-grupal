```python
import pandas as pd
import numpy as np
from google.colab import files

print("--- PASO 1: Sube tu archivo EXCEL 2024 ---")
print("Haz clic en el botón de abajo y selecciona tu archivo: partidos_asistencia_y_precios_2024 (1).xlsx")
uploaded = files.upload()

# Obtener automáticamente el nombre del archivo recién subido
nombre_archivo = list(uploaded.keys())[0]

# Leer las 3 hojas del Excel y unirlas en una sola base limpia
xls = pd.ExcelFile(nombre_archivo)
dfs = []
for sheet in xls.sheet_names:
    df_sheet = pd.read_excel(nombre_archivo, sheet_name=sheet, header=3)
    df_sheet.insert(0, 'Club', sheet)
    dfs.append(df_sheet)

df = pd.concat(dfs, ignore_index=True)

# Asignar puntos automáticamente según el resultado deportivo
def assign_points(row):
    if pd.notnull(row.get('Puntos', None)):
        return row['Puntos']
    res = str(row.get('Resultado', '')).strip().lower()
    if 'victoria' in res:
        return 3
    elif 'empate' in res:
        return 1
    elif 'derrota' in res:
        return 0
    return 0

df['Puntos'] = df.apply(assign_points, axis=1)

print("\n--- PASO 2: ¡Listo! Primeras filas de tu base de datos unificada (2024) ---")
display(df.head(10))

print("\n--- PASO 3: Información general de la base ---")
df.info()

print("\n--- PASO 4: Ejemplo de Tabla Dinámica ---")
pivot_ejemplo = pd.pivot_table(
    df,
    index='Club',
    values=['Precio Más Bajo (CLP)', 'Aforo Autorizado'],
    aggfunc={'Precio Más Bajo (CLP)': 'mean', 'Aforo Autorizado': 'mean'}
).reset_index()
display(pivot_ejemplo)
```
